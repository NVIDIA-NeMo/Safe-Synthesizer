# PII Replacement Plan Preview

Quick scratch notebook for trying the HTML plan preview.

This resolves a plan (auto-discovery or inline) and renders:
- entity tags / replacement scope
- related persona groups
- original sample rows (no replacements applied)

```python
from pathlib import Path

import pandas as pd

from nemo_safe_synthesizer.config.data import DataParameters
from nemo_safe_synthesizer.config.pii_replacement import (
    PiiColumnPlan,
    PiiEntity,
    PiiPersona,
    PiiReplacementPlan,
    ReplacePiiConfig,
)
from nemo_safe_synthesizer.pii_replacer import TabularPiiReplacer, build_plan_preview
```

## Load sample data

Uses a small slice of `patient_events.csv` so discovery stays fast.

```python
# DATA_PATH = Path("/root/datasets/patient_events.csv")
DATA_PATH = Path("/root/datasets/telco_churn.csv")
df = pd.read_csv(DATA_PATH).head(40)
df.head()
```

## Auto-discovery preview

Return the preview object as the last expression in a cell to render HTML automatically.

```python
# data_config = DataParameters(
#     group_training_examples_by="patient_id",
#     order_training_examples_by="event_id",
# )
data_config = DataParameters()
config = ReplacePiiConfig(
    replacement={"locale": "en_US"},
    person={"backend": "faker"},  # faker keeps this notebook self-contained
)

preview = build_plan_preview(
    df,
    config,
    data_config=data_config,
    num_rows=5,
    include_unaffected=True,
)
preview
```

Same preview via the replacer API, with unaffected columns hidden:

```python
replacer = TabularPiiReplacer(config, data_config=data_config)
preview_focused = replacer.preview_plan(df, num_rows=5, include_unaffected=False)
preview_focused.display()
```

## Inspect the resolved plan object

```python
plan = preview.plan
print("group_key:", plan.group_key)
print("personas:", list(plan.identified_personas))
print("advisories:", preview.advisories)
print()
for col, spec in plan.columns.items():
    print(
        f"{col:20s} entity={spec.entity_type}  persona={spec.persona}  "
        f"pattern={spec.pattern}  coverage={spec.dominant_pattern_coverage}"
    )
```

## Inline plan preview

Useful for checking how a hand-written plan renders before applying it.

```python
inline_plan = PiiReplacementPlan(
    group_key="patient_id",
    identified_personas={
        "patient": PiiPersona(gender="sex", ethnic_background="race"),
        "provider": None,
    },
    columns={
        "patient_id": PiiColumnPlan(entity_type=PiiEntity.unique_identifier),
        "first_name": PiiColumnPlan(entity_type=PiiEntity.first_name, persona="patient"),
        "last_name": PiiColumnPlan(entity_type=PiiEntity.last_name, persona="patient"),
        "date_of_birth": PiiColumnPlan(entity_type=PiiEntity.date_of_birth),
        "provider_name": PiiColumnPlan(entity_type=PiiEntity.full_name, persona="provider"),
        "notes": PiiColumnPlan(entity_type=PiiEntity.free_text),
    },
)

inline_preview = build_plan_preview(
    df,
    ReplacePiiConfig(replacement_plan=inline_plan, person={"backend": "faker"}),
    data_config=data_config,
    num_rows=3,
    include_unaffected=False,
)
inline_preview
```
